```python
# Set default catalog and schema
catalog = dbutils.widgets.get("catalog")
schema = dbutils.widgets.get("schema")
spark.sql(f"USE CATALOG {catalog}")
spark.sql(f"CREATE SCHEMA IF NOT EXISTS {schema}")
spark.sql(f"USE SCHEMA {schema}")
```

```python
spark.read.table("samples.nyctaxi.trips")
```

```python
df_transaction_summary = spark.read.table("samples.bakehouse.sales_transactions")

from pyspark.sql import functions as F

df_agg = df_transaction_summary.groupBy("product", "paymentMethod").agg(
    F.sum("totalPrice").alias("totalprice"),
    F.sum("quantity").alias("quantity")
)
# display(df_agg.select("product", "paymentMethod", "totalprice", "quantity"))
df_agg.write.mode("overwrite").saveAsTable(f"{catalog}.{schema}.sales_summary")

```
