```python
from pyspark.sql.functions import * 

df = spark.read.table("samples.bakehouse.sales_franchises")
df2 = df.groupBy("size").agg(

    count("supplierID").alias("NoOfSupplier")
)
# display(df2)

df2.write.mode("overwrite").saveAsTable(f"{catalog}.{schema}.sales_franchises")
```
