```python

from pyspark.sql.functions import *
df = spark.read.table("samples.bakehouse.sales_customers")

customer_360 = (
    df.groupBy("country")
      .agg(
          count("*").alias("num_customers"),
          sum(when(df.gender == "male", 1).otherwise(0)).alias("num_male_customers"),
          sum(when(df.gender == "female", 1).otherwise(0)).alias("num_female_customers")
      )
)

customer_360.write.mode('overwrite').saveAsTable(f"{catalog}.{schema}.customer_360")
```
